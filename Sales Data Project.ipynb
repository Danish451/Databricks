```sql
%python
from pyspark.sql.functions import *
```

```sql
%python
df = spark.read.csv("/Volumes/dev/demo/raw/sales.csv", header=True)
```

```sql
%python
df.display()
```

```sql
%python
print(df.count())          # total number of rows
print(len(df.columns))     # total number of columns
```

```sql
%python
df1 = df.dropDuplicates(["customer_id"]).dropna()
df1.display()
df1.count()                 # Distinct customers
```

```sql
%python
df2 = df1.dropDuplicates(["product_id"]).dropna()
df2.display()
df2.count()             # Distinct Products
```

```sql
%python
df3 = df2.dropDuplicates()
df3.display()
```

```sql
%python
df4 = df3.dropDuplicates(["transaction_id"])
df4.display()
```

```sql
%python
df5 = df4.dropna()
df5.display()
```

```sql
%python
df6 = df5.withColumnRenamed("order_id", "orderId")
df6.display()
```

```sql
%python
df6 = df5.withColumn("order_date", col("order_date").cast("date"))
df6.display()
```

```sql
%python
df7 = df6.withColumn("metadatafilename", col("_metadata.file_name")).withColumn("metadatafilepath", col("_metadata.file_path")).withColumn("timestamp", current_timestamp())

df7.display()
```

```sql
%python
df7.write.mode("overwrite").saveAsTable("dev.demo.sales_table")
```
