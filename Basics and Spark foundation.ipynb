```sql
CREATE VIEW dev.demo.orderpriority_count AS (
    SELECT o_orderpriority, COUNT(*) AS count
    FROM samples.tpch.orders 
    GROUP BY o_orderpriority
    ORDER BY count DESC
)
```

```sql
SELECT * FROM dev.demo.orderpriority_count;
```

```sql
CREATE OR REPLACE FUNCTION dev.demo.datamasking(x STRING)
RETURNS STRING
RETURN CONCAT(REPEAT("*", LENGTH(x)-2), RIGHT(x, 2));
```

```sql
SELECT dev.demo.datamasking('Danish@gmail.com')
```
